```python
import numpy as np
import matplotlib.pyplot as plt
import os
import os.path as osp

root_path = 'path-to-dataset'
dataset_dirs = ['CA', 'GBA', 'GLA', 'SD']
dataset_fns = ['flowca.npz', 'flowgba.npz', 'flowgla.npz', 'flowsd.npz']
dataset_adjs = []
```

```python
dataset_fp = osp.join(root_path, dataset_dirs[3], dataset_fns[3])
data = np.load(dataset_fp)['data'][...,:1]
T, N, C = data.shape
print(T, N, C)
data = data.transpose(1, 2, 0)
```

```python
selected_data = data[336, 0, :]
```

```python
def plot_time_series(data, title='Time Series', xlabel='Time', ylabel='Value'):
    plt.figure(figsize=(15, 5))
    plt.plot(data, label='Traffic Flow')
    plt.title(title)
    plt.xlabel(xlabel)
    plt.ylabel(ylabel)
    plt.legend()
    plt.grid()
    plt.show()
plot_time_series(selected_data)
```
