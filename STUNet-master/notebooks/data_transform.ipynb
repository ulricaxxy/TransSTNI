## Necessary part, run before your experiment

```python
import numpy as np
import os

dataset = 'CA'
data_dir = 'path-to-dataset'
```

```python
Traffic = np.load(os.path.join(data_dir, dataset, f'flow{dataset.lower()}.npz'))['data'][...,:1]
edge_index = np.load(os.path.join(data_dir, dataset, 'edge_index.npy'))
edge_attr = np.load(os.path.join(data_dir, dataset, 'edge_attr.npy'))
```

```python
Traffic.tofile(os.path.join(data_dir, dataset, f'full{dataset.lower()}.dat'))
edge_index.tofile(os.path.join(data_dir, dataset, 'edge_index.dat'))
edge_attr.tofile(os.path.join(data_dir, dataset, 'edge_attr.dat'))
```
